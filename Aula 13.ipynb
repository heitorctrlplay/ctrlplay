```python
class Veiculo:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo

    def mostrar_detalhes(self):
        print(f"Marca: {self.marca}, Modelo: {self.modelo}")

class Carro(Veiculo):
    def __init__(self, marca, modelo, portas):
        super().__init__(marca, modelo)
        self.portas = portas
```

```python
class Jogador():
    vida = 100 # atributo público
    __ataque =  35 # atributo privado a classe Jogador
    defesa = 25 # atributo privado a classe Jogador
    def __init__(self):
        print(self.vida)
        print(self.__ataque)
        print(self.defesa)
    def setAtaque(self, a):
        self.__ataque = a
    def getAtaque(self):
        return self.__ataque

#Programa Principal
jogador = Jogador()
print(jogador.vida)
jogador.setAtaque(40)
print(jogador.getAtaque())
# Erro, __ataque é um atributo privado, somente chamado pela classe
```

```python
#Polimorfismo:

class MeioTransporte():
    def __init__(self, tipo, peso):
        self.tipo = tipo
        self.peso = peso

    def mover(self):
        pass

class Rodoviario(MeioTransporte):
    def mover(self):
        return 'Corre'

class Aereo(MeioTransporte):
    def mover(self):
        return 'Voa'

class Maritimo(MeioTransporte):
    def mover(self):
        return 'Navega'

#Programa principal
c = Rodoviario('carro', '1 ton')
print(c.tipo, c.peso, c.mover())

a = Aereo('avião', '50 ton')
print(a.tipo, a.peso, a.mover())

m = Maritimo('barco', '30 ton')
print(m.tipo, m.peso, m.mover())
```
