```python
def pergunta_Numero():
    numero = 1
    while True:
        try:
            val = int(input('Digite um número inteiro: '))
        except:
            print('Parece que você não digitou um número inteiro')
            continue
        else:
            print('Muito bem!')
            break
        finally:
            print('Tentativa número: {}'.format(numero))
            numero += 1
    print(val)

#Programa principal
pergunta_Numero()
```

```python
try:
    a = int(input())

except:
    ('Parece que você não digitou um número inteiro')

else:
    for num in range(1, a + 1):
        print('{} {} {}'.format(num, num**2, num**3))
```

```python
def pergunta_Senha():
    senha_gravada = 2002
    while True:
        try:
            a = int(input())
        except: 
            print('Senha Invalida')
            continue
        else:
            if a == senha_gravada:
                print('Acesso Permitido')
                break

            else: 
                print('Senha Invalida')

pergunta_Senha()
```
